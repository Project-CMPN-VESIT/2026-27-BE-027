```python
import pandas as pd

fam = pd.read_csv("/content/drive/MyDrive/RiceData/rice4k_geno_no_del.fam",
                  sep=r'\s+', header=None,
                  names=['FID','IID',' PAT','MAT','SEX','PHENO'])

# ID prefix breakdown
def get_prefix(x):
    if str(x).startswith('IRIS'): return 'IRIS_313 (3KRG)'
    elif str(x).startswith('B'): return 'B___ (RiceVarMap)'
    elif str(x).startswith('CX'): return 'CX___ (RiceVarMap)'
    elif str(x).startswith('HP'): return 'HP___ (RiceVarMap)'
    elif str(x).startswith('GP'): return 'GP___ (RiceVarMap)'
    else: return 'Other: ' + str(x)[:4]

fam['prefix'] = fam['IID'].apply(get_prefix)
print(fam['prefix'].value_counts())
```

```python
geno_ids = set(fam['IID'].astype(str))

# phenotype file
pheno = pd.read_csv("/content/drive/MyDrive/RiceData/phenos (1).csv")

pheno_ids = set(pheno['id_name'].astype(str))

# intersection
common_ids = geno_ids.intersection(pheno_ids)

print("Total genotype IDs:", len(geno_ids))
print("Total phenotype IDs:", len(pheno_ids))
print("Matched IDs:", len(common_ids))

print("\nFirst 50 matches:")
print(sorted(list(common_ids))[:500])
```

```python
pheno_matched = pheno[pheno['id_name'].isin(common_ids)]

print(pheno_matched.shape)
pheno_matched.head()
```

```python
matched_ids = set(pheno_matched['id_name'])

fam = pd.read_csv(
    "/content/drive/MyDrive/RiceData/rice4k_geno_no_del.fam",
    sep=r"\s+",
    header=None,
    names=['FID','IID','PAT','MAT','SEX','PHENO']
)

geno_matched = fam[fam['IID'].isin(matched_ids)]

print(geno_matched.shape)
geno_matched.head()
```

```python
import pandas as pd

matched_ids = set(matched_ids)

chunks = []

for chunk in pd.read_csv(
    "/content/drive/MyDrive/RiceData/rice_ml.raw",
    sep=r"\s+",
    chunksize=500
):
    temp = chunk[chunk["IID"].isin(matched_ids)]

    if len(temp) > 0:
        chunks.append(temp)

geno_filtered = pd.concat(chunks)

print(geno_filtered.shape)
```

```python
import pandas as pd

pheno = pd.read_csv("/content/drive/MyDrive/RiceData/phenos (1).csv")

print(pheno.shape)
print(pheno.columns)
pheno.head()
```

```python
# genotype file (529 x 152631)
geno_filtered = pd.read_csv("/content/drive/MyDrive/RiceData/geno_529.csv")
geno_filtered.shape

# phenotype file (529 x 12)
pheno.shape

final_df = geno_filtered.merge(
    pheno,
    left_on="IID",
    right_on="id_name",
    how="inner"
)

print(final_df.shape)
final_df.head()
```

```python
snp_cols = final_df.columns[6:152631]

X = final_df[snp_cols]

y = final_df["Plant_height"]

print(X.shape)
print(y.shape)
```

Select k best

```python
from sklearn.feature_selection import SelectKBest, f_regression

selector = SelectKBest(
    score_func=f_regression,
    k=5000
)

# Fill NaN values in X with 0 (a common approach for genotype data)
X_imputed = X.fillna(0)

X_selected = selector.fit_transform(X_imputed, y)

print(X_selected.shape)
```

```python
print(final_df.shape)

final_df.head()
```

```python
pheno_cols = [
    'Heading_date',
    'Plant_height',
    'Num_panicles',
    'Num_effective_panicles',
    'Yield',
    'Grain_weight',
    'Spikelet_length',
    'Grain_length',
    'Grain_width',
    'Grain_thickness'
]

# Get the names of the selected SNP columns
selected_snps = X.columns[selector.get_support()]

reduced_df = pd.DataFrame(
    X_selected,
    columns=selected_snps
)

reduced_df = pd.concat(
    [reduced_df, final_df[pheno_cols].reset_index(drop=True)],
    axis=1
)

print(reduced_df.shape)
```

```python
reduced_df.insert(
    0,
    "IID",
    final_df["IID"].values
)

reduced_df.insert(
    1,
    "id_name",
    final_df["id_name"].values
)
```
